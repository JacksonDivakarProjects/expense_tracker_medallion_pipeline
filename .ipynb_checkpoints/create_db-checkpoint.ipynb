```python
import duckdb
import pandas as pd
```

```python


# 1. Connect to the local DuckDB file
db_path = "data/02_db/personal_finance.duckdb"
conn = duckdb.connect(db_path)

# 2. Query the raw layer
print("--- Raw Expense Journal ---")
raw_df = conn.execute("SELECT * FROM raw.expense_journal LIMIT 5;").df()
display(raw_df)

```

```python
# 3. Query the staging layer (if Step 3 succeeded after your fix)
print("\n--- Staging Fact Expenses ---")
try:
    staging_df = conn.execute("SELECT * FROM staging.fact_expenses LIMIT 5;").df()
    display(staging_df)
except duckdb.CatalogException:
    print("Staging table does not exist yet. Run main.py to build it.")
```
